## <span style="color:purple">Finding consistency errors with BasicConsistencyChecker</span>

The [estnltk_neural](https://github.com/estnltk/estnltk/tree/main/estnltk_neural) package contains `BasicConsistencyChecker`, which searches for basic consistency errors regarding deprels and feats.

## Prerequisites

*Note: you need to install [estnltk_neural](https://github.com/estnltk/estnltk/tree/main/estnltk_neural) package.*

Make sure that the correct feats extraction is selected in the '*_check_consistency*' function. The difference is that 'feats' is expected to be a dictionary and 'extended_feats' is a string where all feats are separated by a comma. Dictionary version is used in the older dataset layers and 'extended_feats' is intended for newer version in the future.

These examples are from the older dataset.

## Examples how to use BasicConsistencyChecker

The files used in this tutorial can be found also in the [test folder](https://github.com/estnltk/estnltk/tree/devel_1.7/estnltk_neural/estnltk_neural/tests/taggers/morphosyntax/test_files_basic_consistency).

```python
from estnltk_neural.taggers import BasicConsistencyChecker
from estnltk import Text
from estnltk.converters import json_to_text, text_to_json
import os
```

```python
tagger = BasicConsistencyChecker(
                        morphosyntax_layer = "v172_stanza_syntax",
                         output_layer="morphosyntax_consistency"
                        )
```

### Examples for all the rules in the tagger

```python
# no errors, empty layer

s = json_to_text(file = "sentence_510950.json")

tagger.tag(s)
print(s.text)
s["morphosyntax_consistency"]
```

```python
# nsubj but not 'nom'/'part'

s = json_to_text(file = "sentence_3.json")

tagger.tag(s)
print(s.text)
s["morphosyntax_consistency"]
```

```python
# obj but not 'nom'/'part'/'gen'
# xcomp and feats is not empty

s = json_to_text(file = "sentence_5.json")

tagger.tag(s)
print(s.text)
s["morphosyntax_consistency"]
```

```python
# advcl and feats is not empty

s = json_to_text(file = "sentence_16.json")

tagger.tag(s)
print(s.text)
s["morphosyntax_consistency"]
```
