## <span style="color:purple">Morphological and Stanza analysis with MorphoSyntaxTagger</span>

The [estnltk_neural](https://github.com/estnltk/estnltk/tree/main/estnltk_neural) package contains `MorphoSyntaxTagger`, which provides VabamorfWithBertTagger's morphological analysis together with Stanza Syntax analysis.

## Prerequisites

*Note: you need to install [estnltk_neural](https://github.com/estnltk/estnltk/tree/main/estnltk_neural) package.*

Note that the Stanza model required by the tagger is not distributed with the estnltk\\_neural package.

MorphoSyntaxTagger uses:
- existing morph layer or VabamorfWithBertTagger (or other tagger for morph analysis layer)
- MorphExtendedTagger
- StanzaSyntaxTagger

For morph_analysis layer, either an existing layer name is provided with "morph_analysis_layer" parameter or the tagger is passes with "morph_analysis_tagger" parameter. There is no need to provide the tagger if morph_analysis layer already exists. If tagger is given, then a new morph_analsysi layer will be created regardless and "morph_analysis_layer" value will be overwritten with the morph_analysis_tagger's output_layer.

## Examples how to use MorphoSyntaxTagger

```python
%load_ext autoreload
%autoreload 2
```

```python
from estnltk_neural.taggers import VabamorfWithBertTagger
from estnltk.taggers import MorphExtendedTagger
from estnltk_neural.taggers import StanzaSyntaxTagger
from estnltk_neural.taggers import MorphoSyntaxTagger
from estnltk import Text
```

### Usage without an existing morph_analysis layer

Requires:

- VabamorfWithBertTagger (or other tagger for morph analysis layer)
- MorphExtendedTagger
- StanzaSyntaxTagger

```python
vwbt = VabamorfWithBertTagger(output_layer="morph_with_bert")

met = MorphExtendedTagger(output_layer="morph_extended", input_morph_analysis_layer="morph_with_bert")

depparse_path = "../stanza_treenimine/stanza1.14_ud2.18/model_morph_with_bert.pt"
sst = StanzaSyntaxTagger( output_layer="stanza_syntax", 
                                 input_type="morph_analysis",
                                 input_morph_layer="morph_with_bert",
                                 words_layer="words", 
                                 sentences_layer="sentences", 
                                 resources_path=None, 
                                 depparse_path =depparse_path, 
                                 add_parent_and_children=False, 
                                 use_gpu=False)
```

```python
mst = MorphoSyntaxTagger(morph_analysis_tagger = vwbt,
                        morph_extended_tagger = met,
                         stanza_tagger = sst,
                         output_layer = "morphosyntax"
                        )
```

```python
t3 = Text("""Kas sa tead ka, väike Rilli , et meil aias kasvab lilli? \
    Saal neid mitmesuguseid palistamas aiateid . \
    Koiduajal tõttan ma oma aeda vaatama . \
    Hommikul , kui kastan lilli , siis veel magad, väike Rilli . \
    Tule, Rilli, abiks mulle, tasuks kondi annan sulle !  \
    Lapsed aias tegutsevad , jalgteed puhtaks rehitsevad , siis veel kuldse liivaga kaetakse need hoolega . \
    Pärast puhata on ilus varjupikka põõsa vilus . \
    Ilmad kuumad , palavad , lilled aias näräbuvad. \
    Kastma pead neid hoolega , et nad võiksid kosuda . \
    Kui sul janu , aitab vesi . \
    Samuti ka lillekesi jooda igal hommikul , aega selleks küllalt sul . \
    Kaks lille ilusat on mul , toas kasvatasin endale , nüüd aga suve tulekul viin välja päiksepaistele . \
    Kaks lille ilusat on mul , need kingin varsti emale , ning sünnipäeva hommikul las teevad rõõmu temale . \
    Mesilane - linnukene lendab lillelt lillele , ise imetillukene , aga hoolas töödele. \
    Mesilane - linnukene mett sääl korjab sinule . \
    Mesi magus, lapsukene , maitseb igaühele ! \
    Palju lilli, palju päikest suve läbi meie aias rõõmustamas suurt ja väikest ; marju leiab suuke maias. \
    Tule meile, palun väga, aias tore mängida , tervitan sind lilledega kingin õuna magusa .""")

# create necessary layers before
t3.tag_layer(["words", "sentences"])

# morphosyntax tagging
mst.tag( t3 )
```

```python
t3["sentences"][1].morphosyntax
```

### Usage with an existing morph_analysis layer

Requires:

- morph_analysis layer
- MorphExtendedTagger
- StanzaSyntaxTagger

```python
# still need the tagger to create the necessary layer beforehand
vwbt = VabamorfWithBertTagger(output_layer="morph_with_bert")

met = MorphExtendedTagger(output_layer="morph_extended", input_morph_analysis_layer="morph_with_bert")

depparse_path = "../stanza_treenimine/stanza1.14_ud2.18/model_morph_with_bert.pt"
sst = StanzaSyntaxTagger( output_layer="stanza_syntax", 
                                 input_type="morph_analysis",
                                 input_morph_layer="morph_with_bert",
                                 words_layer="words", 
                                 sentences_layer="sentences", 
                                 resources_path=None, 
                                 depparse_path =depparse_path, 
                                 add_parent_and_children=False, 
                                 use_gpu=False)
```

```python
mst = MorphoSyntaxTagger(
                        morph_analysis_layer = "morph_with_bert",
                        morph_extended_tagger = met,
                         stanza_tagger = sst,
                         output_layer = "morphosyntax"
                        )
```
