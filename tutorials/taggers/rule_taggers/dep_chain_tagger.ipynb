This notebook shows how to configure `DepChainTagger`, define a `PathPattern`, and run the tagger on a sample text. It also explains the most important parts of the output so you can adapt the example to your own texts. The `DepChainTagger` is a tool for identifying and tagging dependency chains in text, which can be useful for various natural language processing tasks. In this notebook, we will go through the steps to set up the tagger, create a pattern to match specific dependency chains, and analyse the results.

The last section will focus on `DepChildTagger`, a related tool that allows tagging child nodes in dependency trees.


# DepChainTagger


````python
import estnltk

from estnltk.taggers import (
    ConditionMode,
    DirectionMode,
    EdgeConstraint,
    NodeConstraint,
    PathPattern,
    ValueCondition,
    NestedValueCondition,
)

from estnltk.taggers import (
    DepChainTagger,
    DepChildTagger,
)

from estnltk_neural.taggers import StanzaSyntaxTagger
````

## Tagger parameters

The current `DepChainTagger` constructor focuses on output control and match filtering. The tagger always reads from the `stanza_syntax` and `sentences` layers internally, and it writes a relation layer named `dep_chains` by default.

| Parameter                      | What it controls                                                                                                      |
| ------------------------------ | --------------------------------------------------------------------------------------------------------------------- |
| **`patterns`**                 | A tuple of `PathPattern` objects that define what the tagger should match.                                            |
| `syntax_layer`                 | Name of the syntax layer used as input. Default: `"stanza_syntax"`                                                    |
| `sentences_layer`              | Name of the sentence layer used to split the text. Default: `"sentences"`                                             |
| `output_layer`                 | Name of the relation layer created by the tagger. Default: `"dep_chains"`                                             |
| `output_span_names`            | Tuple of span names to include in the output. Default: all span names from the patterns                               |
| `output_attributes`            | Extra attributes written to the output layer. Tagger adds role constraints if `include_pattern_constraints` is True.  |
| `include_pattern_constraints`  | Whether to include the pattern constraints as attributes in the output spans/relations.                               |
| `max_matches_per_sentence`     | Maximum number of matches to keep per sentence.                                                                       |
| `allow_role_node_overlap`      | Whether the same node may fill more than one role in a pattern.                                                       |
| `max_total_matches`            | Upper limit for all matches across the entire text.                                                                   |
| `sentence_match_dedup_mode`    | Deduplication inside one sentence: `none`, `exact`, or `role_based`.                                                  |
| `global_dedup_mode`            | Deduplication across the whole text: `none`, `exact`, or `role_based`.                                                |

**Everything except the `patterns` is optional and has a default value.** The deduplication modes and their implications are explained in more detail below.

When you create a `DepChainTagger` instance with `include_pattern_constraints=True` and define `output_attributes`, the tagger will check if any of the user-defined attributes will conflict with the pattern constraints. If there is a conflict, the tagger will raise an error to prevent overwriting important information in the output layer.

## Constructing a tagger with a simple pattern


A tagger is configured by passing the desired parameters to the constructor. The most important parameter is `patterns`: a tuple of `PathPattern` objects that define what the tagger will look for. A `PathPattern` consists of:

- `node_steps` (a sequence of `NodeConstraint` objects) and
- `edge_steps` (a sequence of `EdgeConstraint` objects); in typical linear patterns `len(edge_steps) == len(node_steps) - 1`.

Each `NodeConstraint` assigns a `role` (this role name is used in the output relation) and can include attribute or feature conditions; each `EdgeConstraint` specifies direction and hop limits. The matcher traverses the `stanza_syntax` tree following the pattern and emits matches as relations (default layer `dep_chains`) where roles point to the matched syntax nodes.

Below is a simplified overview of the workflow:

```
(NodeConstraints, EdgeConstraints) --> PathPattern (with roles) --> DepChainTagger --> Output relation layer with matches
```


Let's start with a simple pattern that finds a root and its direct children in the syntax layer. We will define a pattern with two nodes: the first node is the root of the sentence, and the second node is any child of the root. We will label the root as `root` and the child as `first_level_child` in the output relation layer. The edge constraint will specify that they can look only one step away from the root.


````python
root_pattern = PathPattern(
    name="root_and_children",
    node_steps=(
        NodeConstraint(
            role="root",
            attribute_conditions={
                "deprel": ValueCondition(mode=ConditionMode.EXACT, value="root"),
                # "text": ValueCondition(mode=ConditionMode.WILDCARD),
                "upostag": ValueCondition(mode=ConditionMode.WILDCARD),
            },
        ),
        NodeConstraint(
            role="first_level_child",
            attribute_conditions={
                "deprel": ValueCondition(mode=ConditionMode.EXACT, value="obl")
            },
        ),
    ),
    edge_steps=(
        EdgeConstraint(
            direction=DirectionMode.DOWN,
            min_hops=1,
            max_hops=1,
        ),
    ),
)
````

Now, let's instantiate the tagger with this pattern and see how it works on our sample text.


````python
verb_tagger = DepChainTagger(patterns=(root_pattern,), include_pattern_constraints=True)
````

````python
sample_text = "1990. aasta kuumal suvel vaatas Bureau Veritas Estline'i omanduseks saanud laeva uuesti üle."
text_obj = estnltk.Text(sample_text)
text_obj.tag_layer("morph_extended")
stanza_syntax_tagger.tag(text_obj)

# Run the tagger
verb_tagger.tag(text_obj);
````

Let's check the Text object after tagging.


````python
display(text_obj)
````

There is a new relation layer named `dep_chains` that contains the matches found by the tagger. Each match is a relation with the roles defined by the pattern (for example, `root` and `first_level_child` in the simple pattern above). The relation also contains several attributes:

- One attribute for each role defined in the pattern, pointing to the matched node for that role.
- `pattern_name`: the name of the pattern that produced the match.
- `matched_text`: the surface text covered by the matched nodes.

If you enabled `include_pattern_constraints=True` when constructing the tagger, as we did here, additional attributes derived from the pattern's constraints (prefixed with the role name) will also be included in the output relation. The exact number and content of matches depend on the quality of the syntax parse and the Stanza model version used in your environment.

The output layer is a relation layer (default name `dep_chains`), so each match is represented as a relation whose role fields point to the matched syntax nodes/spans (e.g. `root`, `first_level_child`). Each relation also contains standard attributes: `pattern_name` (the producing pattern) and `matched_text` (surface text covered by the matched nodes). If `include_pattern_constraints=True` was set when constructing the tagger, additional attributes derived from the pattern constraints are included, prefixed with the role name (for example, `verb_upostag`).

In the example above:

- The `root` attribute points to the root's `role` we defined in the pattern.

- The `matched_text` attribute shows the text covered by the matched nodes.

- The `first_level_child_upostag` attribute shows the constraint defined for the `first_level_child` role in the pattern.


## Going in-depth on patterns and tagger configuration


### Path patterns


The `PathPattern` class is the core of the pattern-matching system. It defines a sequence of node and edge constraints (`NodeConstraint` and `EdgeConstraint` respectively) that describe the structure we want to match in the dependency tree. Each node constraint specifies a role and conditions on the node's attributes or features, while each edge constraint specifies the direction and hop limits for traversing the tree.

The tagger can take multiple `PathPattern` objects, allowing you to search for different structures in the same run. Each pattern can have a name, and the output relations will indicate which pattern produced each match.

Let's look at the two main types of constraints you can use: `NodeConstraint` and `EdgeConstraint`.


### NodeConstraints


`NodeConstraint` objects specify conditions on the nodes in the pattern. They can check for specific attribute values, the presence of certain attributes, or more complex nested checks using `NestedValueCondition`.

**Key parameters**

- `role`: The role name for the node in the output relation layer.
- `attribute_conditions`: A dictionary where keys are scalar attribute names and values are condition objects such as `ValueCondition`. Use this for ordinary node attributes like `upostag`, `deprel`, `lemma`, `id`, or `head`. Example: `{"upostag": ValueCondition(mode=ConditionMode.EXACT, value="V")}`
- `nested_attribute_conditions`: A dictionary where keys are names of nested or dictionary-like attributes (for example, `feats`) and values are `NestedValueCondition` objects used to test those nested structures. Example: `{"feats": NestedValueCondition(required={"Number": "Sing"})}`
- `extra_predicates`: An optional sequence of callables `predicate(node) -> bool` that perform arbitrary checks on candidate nodes. All predicates must return `True` for the node to satisfy the constraint.

**Semantics & tips**

- Use `attribute_conditions` for scalar attributes (simple equality/regex/membership checks via `ValueCondition`)
- Use `nested_attribute_conditions` for nested/dict-like attributes (use `NestedValueCondition` to express required/forbidden sub-structures without needing to match the entire structure).
- An empty `NodeConstraint` is most efficient; use `WILDCARD` to document intent when you want any node value, but also record the constraint (useful when `include_pattern_constraints=True`).

---


Let's create a simple `NodeConstraint` that matches a verb whose `feats` are `s`. We can use `attribute_conditions` for the `upostag` to specify that it must be a verb (upostag `V`), and we can use `nested_attribute_conditions` for the `feats` to specify that it must be third person singular (`ps3` and `sg`).


````python
pattern_1 = PathPattern(
    name="node_constraint",
    node_steps=(
        NodeConstraint(
            role="verb_s",
            attribute_conditions={
                "upostag": ValueCondition(mode=ConditionMode.EXACT, value="V")
            },
            nested_attribute_conditions={
                "feats": NestedValueCondition(
                    mode=ConditionMode.EXACT,
                    required={
                        "ps3": "ps3",
                        "sg": "sg",
                    },
                    allow_extra_keys=True,  # allowing extra keys means we only require the presence of ps3 and sg, but other features can also be present without causing a mismatch.
                    # An example of this parameter's usage is further down in the notebook.
                ),
            },
        ),
    ),
    edge_steps=(),
)
````

````python
# Tagger instantiation
dct = DepChainTagger(patterns=(pattern_1,))

# Sample text with syntax layer
sample_text = "1990. aasta kuumal suvel vaatas Bureau Veritas Estline'i omanduseks saanud laeva uuesti üle."
text_obj = estnltk.Text(sample_text)
text_obj.tag_layer("morph_extended")
stanza_syntax_tagger.tag(text_obj)

# Run the tagger
dct.tag(text_obj);
````

````python
display(text_obj["dep_chains"])
text_obj["dep_chains"].display()
````

We see that the word "vaatas" fits the constraints we defined: it has `upostag` `V` and its `feats` include `ps3` and `sg`.

---


### EdgeConstraints


`EdgeConstraint` objects specify the relationship between consecutive `NodeConstraint` steps in a `PathPattern`. Edge attribute checks test the dependency relation connecting the two nodes, not a node's own attributes.

**Key parameters**

- `direction`: `DirectionMode.DOWN` or `DirectionMode.UP`. `DOWN` moves from a node to its child(ren); `UP` moves to the parent. Direction controls how the matcher traverses the `stanza_syntax` tree.
- `min_hops`, `max_hops`: Integers controlling how many dependency hops may be traversed for this edge. Use `min_hops=max_hops=1` for a single immediate relation; larger ranges allow skipping intermediate nodes.

**Semantics & tips**

- Prefer precise `min_hops`/`max_hops` and `attribute_conditions` to improve matching speed and reduce spurious matches.
- Longer hops will skip intermediate nodes and check only the first and last nodes in the path.
- When direction is set to `BOTH` and hop range is greater than 1, the matcher will check in both directions unidirectionally. This means that it will not check for a path that goes down and then up, or up and then down; it will only check for paths that go in one direction (either all down or all up) within the specified hop range. For mixed-direction paths, you need to explicitly define edge constraints for each direction.
- `DepChildTagger` enforces that all edges are `DirectionMode.DOWN` (it matches child-only relations); using `UP` edges will raise a `ValueError` during validation.

---


Let's make the pattern more complex by changing it from matching a node to matching a chain of two nodes: a root and its child. We will need to define two `NodeConstraint` objects (one for the root and one for the child) and an `EdgeConstraint` to connect them.

We create three patterns with the same node constraints but different edge constraints to illustrate how edge conditions affect the matches. The first pattern requires an immediate child (`min_hops=1, max_hops=1`), the second allows descendants at up to 4 levels of separation (`min_hops=1, max_hops=4`), and the third allows descendants that are not direct children but are within 4 hops (`min_hops=2, max_hops=4`).


````python
pattern_1 = PathPattern(
    name="root_and_obl_children",
    node_steps=(
        NodeConstraint(
            role="root",
            attribute_conditions={
                "text": ValueCondition(mode=ConditionMode.EXACT, value="vaatas")
            },
        ),
        NodeConstraint(
            role="child",
            attribute_conditions={
                "deprel": ValueCondition(mode=ConditionMode.EXACT, value="obl")
            },
        ),
    ),
    edge_steps=(
        EdgeConstraint(
            direction=DirectionMode.DOWN,
            min_hops=1,
            max_hops=4,
        ),
    ),
)

pattern_2 = PathPattern(
    name="root_and_obl_direct_children",
    node_steps=(
        NodeConstraint(
            role="root",
            attribute_conditions={
                "text": ValueCondition(mode=ConditionMode.EXACT, value="vaatas")
            },
        ),
        NodeConstraint(
            role="child",
            attribute_conditions={
                "deprel": ValueCondition(mode=ConditionMode.EXACT, value="obl")
            },
        ),
    ),
    edge_steps=(
        EdgeConstraint(
            direction=DirectionMode.DOWN,
            min_hops=1,
            max_hops=1,
        ),
    ),
)

pattern_3 = PathPattern(
    name="root_and_obl_not_direct_children",
    node_steps=(
        NodeConstraint(
            role="root",
            attribute_conditions={
                "text": ValueCondition(mode=ConditionMode.EXACT, value="vaatas")
            },
        ),
        NodeConstraint(
            role="child",
            attribute_conditions={
                "deprel": ValueCondition(mode=ConditionMode.EXACT, value="obl")
            },
        ),
    ),
    edge_steps=(
        EdgeConstraint(
            direction=DirectionMode.DOWN,
            min_hops=2,
            max_hops=4,
        ),
    ),
)
````

````python
# Tagger instantiation
dct = DepChainTagger(patterns=(pattern_1, pattern_2, pattern_3))

# Sample text with syntax layer
sample_text = "1990. aasta kuumal suvel vaatas Bureau Veritas Estline'i omanduseks saanud laeva uuesti üle."
text_obj = estnltk.Text(sample_text)
text_obj.tag_layer("morph_extended")
stanza_syntax_tagger.tag(text_obj)

# Run the tagger
dct.tag(text_obj);
````

````python
display(text_obj["dep_chains"])
text_obj["dep_chains"].display()
````

We see that the first pattern matches all children of the root node that satisfy the specified node constraints. The second pattern matches only the immediate children of the root, while the third pattern matches only the more distant descendants (not direct children) that still satisfy the node constraints. This illustrates how edge constraints can control the structural relationships between matched nodes in a `PathPattern`.

### Condition types


There are two condition types you will see in pattern definitions: `ValueCondition` and `NestedValueCondition`.

`ValueCondition`

`ValueCondition` is the default choice for simple scalar attributes such as `upostag`, `deprel`, `lemma`, `id`, or `head`. Use it whenever the attribute you are checking is a single value (e.g., string, number). `ValueCondition` supports the standard condition modes: `EXACT`, `NEGATION`, `WILDCARD`, `MEMBERSHIP`, `NOT_MEMBERSHIP`, and `REGEX` (for text-based pattern matching).

Typical `ValueCondition` examples:

- check that a node is a verb: `{"upostag": ValueCondition(mode=ConditionMode.EXACT, value="V")}`
- allow verbs or nouns: `{"upostag": ValueCondition(mode=ConditionMode.MEMBERSHIP, value=["V", "S"])}`
- match by regular expression: `{"deprel": ValueCondition(mode=ConditionMode.REGEX, value="^obj.*")}`

`NestedValueCondition`

`NestedValueCondition` is intended for nested, dictionary-like or collection-valued attributes — most commonly the morphological `feats` map. Instead of comparing a single scalar, it lets you assert the presence or absence of sub-keys and sub-values using `required` and `forbidden` specifications (and related options such as `allow_extra_keys`). `NestedValueCondition` is designed for structured matching and does not provide scalar `REGEX` semantics; use `ValueCondition` for regex checks on simple attributes.

Typical `NestedValueCondition` examples:

- require a singular noun inside `feats`: `{"feats": NestedValueCondition(required={"Number": "Sing"})}`
- require several features and forbid another: `NestedValueCondition(required={"Number": "Pl"}, forbidden={"Case": "Gen"})`

Practical rule of thumb

- Use `ValueCondition` for simple, direct attributes on nodes or edges (scalar values, regex, membership checks).
- Use `NestedValueCondition` for nested/dict-like attribute information (for example, `feats`) when you need to assert sub-keys or sub-values.

This separation keeps patterns readable and prevents mixing scalar and nested matching semantics.


### Condition modes for condition types


The meaning of a condition mode depends on the condition class that uses it.

#### `ValueCondition` for scalar attributes

`ValueCondition` is the right choice for simple attributes that hold one value, such as `upostag`, `lemma`, `deprel`, `id`, or `head`. Here, the mode controls how a single actual value is compared with a single expected value.

- `EXACT`: the actual value must be equal to the expected value.
- `NEGATION`: the actual value must be different from the expected value.
- `WILDCARD`: any actual value is accepted; the expected value must be `None`.
- `MEMBERSHIP`: the actual value must be one of the values in the expected iterable.
- `NOT_MEMBERSHIP`: the actual value must not be one of the values in the expected iterable.
- `REGEX`: the text form of the actual value must match the expected regular expression.

Typical examples:

- `upostag=ValueCondition(mode=ConditionMode.EXACT, value="V")` matches only verbs.
- `upostag=ValueCondition(mode=ConditionMode.MEMBERSHIP, value=["V", "S"])` matches verbs and nouns.
- `deprel=ValueCondition(mode=ConditionMode.REGEX, value="^obj.*")` matches dependency labels that start with `obj`.

#### `NestedValueCondition` for nested or dictionary-like attributes

`NestedValueCondition` is used for attributes whose values are nested structures, most commonly the morphological `feats` map. Instead of comparing one scalar value, it compares nested content such as dictionaries, lists, tuples, sets, or combinations of them.

The modes are interpreted differently here:

- `EXACT`: the actual nested structure must contain the required content, and it must not violate the forbidden content. If `allow_extra_keys=False`, no additional keys are allowed beyond those specified by the condition.
- `NEGATION`: the actual nested structure must not match the required content, and it must also avoid the forbidden content.
- `WILDCARD`: any nested value is accepted.
- `MEMBERSHIP`: at least one candidate in `required` must be found somewhere in the nested structure; `forbidden` still blocks a match.

Typical examples:

- `nested_attribute_conditions={"feats": NestedValueCondition(mode=ConditionMode.EXACT, required={"Number": "Sing"})}` matches nodes whose `feats` contain a singular number.
- `nested_attribute_conditions={"feats": NestedValueCondition(mode=ConditionMode.EXACT, required={"Case": "Gen"}, forbidden={"Number": "Plur"})}` matches nodes whose `feats` contain genitive case and do not contain plural number.

#### Practical rule of thumb

- Use `ValueCondition` for ordinary scalar values.
- Use `NestedValueCondition` for nested, dictionary-like, or collection-valued attributes.

The same `ConditionMode` name can therefore behave differently depending on the condition class: in `ValueCondition`, it compares one scalar value. At the same time, in `NestedValueCondition`, it evaluates the shape and content of a nested structure.

This separation keeps pattern definitions readable and makes the intended matching behaviour explicit.


Let's take the example above and modify the pattern to use different condition modes to see how they affect matches and the output layer.


#### Using `EXACT` mode


First is the condition mode `EXACT` that we have already seen. There are two patterns:

- One that matches only verbs with `upostag = V`.
- The other matches only verbs with `upostag = V` and third person singular `feats` (`ps3` and `sg`).


````python
# Pattern definition using EXACT condition mode
v_pattern_exact = PathPattern(
    name="exact_verb_pattern",
    node_steps=(
        NodeConstraint(
            role="verb",
            attribute_conditions={
                "upostag": ValueCondition(mode=ConditionMode.EXACT, value="V")
            },
        ),
    ),
    edge_steps=(),
)
v_pattern_n_exact = PathPattern(
    name="exact_verb_pattern_nested",
    node_steps=(
        NodeConstraint(
            role="verb",
            attribute_conditions={
                "upostag": ValueCondition(mode=ConditionMode.EXACT, value="V")
            },
            nested_attribute_conditions={
                "feats": NestedValueCondition(
                    mode=ConditionMode.EXACT,
                    required={
                        "ps3": "ps3",
                        "sg": "sg",
                    },
                    allow_extra_keys=True,
                )
            },
        ),
    ),
    edge_steps=(),
)
````

````python
# Tagger instantiation
verb_tagger = DepChainTagger(patterns=(v_pattern_exact, v_pattern_n_exact))

# Sample text with syntax layer
sample_text = "1990. aasta kuumal suvel vaatas Bureau Veritas Estline'i omanduseks saanud laeva uuesti üle."
text_obj = estnltk.Text(sample_text)
text_obj.tag_layer("morph_extended")
stanza_syntax_tagger.tag(text_obj)

# Run the tagger
verb_tagger.tag(text_obj);
````

````python
display(text_obj["dep_chains"])
text_obj["dep_chains"].display()
````

The simpler (non-nested-condition) pattern matches both _vaatas_ and _saanud_. In contrast, the more complex pattern with nested conditions matches only _vaatas_ because it is the only verb that has `feats` indicating third person singular.


#### Using `NEGATION` mode


Second is the condition mode `NEGATION`. There are two patterns:

- One that matches any node that is not a verb (negating `upostag = V`).
- The other matches any node that is not a verb with third person singular `feats` (negating both `upostag = V` and the nested `feats` condition).


````python
# Pattern definition using NEGATED condition mode
v_pattern_negated = PathPattern(
    name="negation_verb_pattern",
    node_steps=(
        NodeConstraint(
            role="not_verb",
            attribute_conditions={
                "upostag": ValueCondition(mode=ConditionMode.NEGATION, value="V")
            },
        ),
    ),
    edge_steps=(),
)
v_pattern_n_negated = PathPattern(
    name="negation_verb_pattern_nested",
    node_steps=(
        NodeConstraint(
            role="not_verb",
            nested_attribute_conditions={
                "feats": NestedValueCondition(
                    mode=ConditionMode.NEGATION,
                    required={
                        "ps3": "ps3",
                        "sg": "sg",
                    },
                    allow_extra_keys=True,
                )
            },
        ),
    ),
    edge_steps=(),
)
````

````python
# Tagger instantiation
verb_tagger = DepChainTagger(patterns=(v_pattern_negated, v_pattern_n_negated))

# Sample text with syntax layer
sample_text = "1990. aasta kuumal suvel vaatas Bureau Veritas Estline'i omanduseks saanud laeva uuesti üle."
text_obj = estnltk.Text(sample_text)
text_obj.tag_layer("morph_extended")
stanza_syntax_tagger.tag(text_obj)

# Run the tagger
verb_tagger.tag(text_obj);
````

````python
display(text_obj["dep_chains"])
text_obj["dep_chains"].display()
````

In `NEGATION` mode, we get matches for all nodes whose `upostag` is not `V`. The first pattern matches all non-verbs, while the second pattern matches all nodes that are not with third person singular `feats`, which will include _saanud_ because it does not have `feats` indicating third person singular (the first pattern would exclude it because it is a verb).


#### Using `WILDCARD` mode


The third condition mode is `WILDCARD`, which matches any node regardless of its attributes.


````python
# Pattern definition using WILDCARD condition mode
v_pattern_wildcard = PathPattern(
    name="wildcard_verb_pattern",
    node_steps=(
        NodeConstraint(
            role="all",
            attribute_conditions={
                "upostag": ValueCondition(mode=ConditionMode.WILDCARD, value=None)
            },
        ),
    ),
    edge_steps=(),
)
v_pattern_n_wildcard = PathPattern(
    name="wildcard_verb_pattern_nested",
    node_steps=(
        NodeConstraint(
            role="all",
            attribute_conditions={
                "upostag": ValueCondition(mode=ConditionMode.WILDCARD, value=None)
            },
            nested_attribute_conditions={
                "feats": NestedValueCondition(
                    mode=ConditionMode.WILDCARD, required=None, allow_extra_keys=True
                )
            },
        ),
    ),
    edge_steps=(),
)

v_pattern_wo_wildcard = PathPattern(
    name="no_wildcard_verb_pattern",
    node_steps=(
        NodeConstraint(
            role="all",
            attribute_conditions={
                # Not specifying any conditions means we want to match any value for this attribute, effectively acting as a wildcard without needing to explicitly use WILDCARD.
            },
        ),
    ),
    edge_steps=(),
)
````

````python
# Tagger instantiation
verb_tagger = DepChainTagger(
    patterns=(v_pattern_wildcard, v_pattern_n_wildcard, v_pattern_wo_wildcard),
    include_pattern_constraints=True,
)

# Sample text with syntax layer
sample_text = "1990. aasta kuumal suvel vaatas Bureau Veritas Estline'i omanduseks saanud laeva uuesti üle."
text_obj = estnltk.Text(sample_text)
text_obj.tag_layer("morph_extended")
stanza_syntax_tagger.tag(text_obj)

# Run the tagger
verb_tagger.tag(text_obj);
````

````python
display(text_obj["dep_chains"])
text_obj["dep_chains"].display()
````

As expected, with `WILDCARD` mode, we get matches for all nodes in the syntax layer.


#### Using `MEMBERSHIP` mode


The fourth condition mode, `MEMBERSHIP`, matches nodes whose `upostag` value is either `V` or `S`. There are two patterns:

- One that matches nodes with `upostag` in `["V", "S"]`.
- The other matches nodes with `upostag` in `["V", "S"]`, but also requires that if the node has `feats`, it must include either `ps3` or `sg` (third person or singular).


````python
# Pattern definition using MEMBERSHIP condition mode
v_and_s_pattern_membership = PathPattern(
    name="membership_verb_and_noun_pattern",
    node_steps=(
        NodeConstraint(
            role="verb_or_noun",
            attribute_conditions={
                "upostag": ValueCondition(
                    mode=ConditionMode.MEMBERSHIP, value=["V", "S"]
                )
            },
        ),
    ),
    edge_steps=(),
)
v_and_s_pattern_n_membership = PathPattern(
    name="membership_verb_and_noun_pattern_nested",
    node_steps=(
        NodeConstraint(
            role="verb_or_noun",
            attribute_conditions={
                "upostag": ValueCondition(
                    mode=ConditionMode.MEMBERSHIP, value=["V", "S"]
                )
            },
            nested_attribute_conditions={
                "feats": NestedValueCondition(
                    mode=ConditionMode.MEMBERSHIP,
                    required={
                        "ps3": "ps3",
                        "sg": "sg",
                    },
                    allow_extra_keys=True,
                )
            },
        ),
    ),
    edge_steps=(),
)
````

````python
# Tagger instantiation
verb_tagger = DepChainTagger(
    patterns=(v_and_s_pattern_membership, v_and_s_pattern_n_membership)
)

# Sample text with syntax layer
sample_text = "1990. aasta kuumal suvel vaatas Bureau Veritas Estline'i omanduseks saanud laeva uuesti üle."
text_obj = estnltk.Text(sample_text)
text_obj.tag_layer("morph_extended")
stanza_syntax_tagger.tag(text_obj)

# Run the tagger
verb_tagger.tag(text_obj);
````

````python
display(text_obj["dep_chains"])
text_obj["dep_chains"].display()
````

The first pattern matches all verbs and nouns. The second pattern is almost the same, because all nouns have `sg` in their `feats` if they are singular. However, it excludes _saanud_ as it does not have `feats` indicating third person nor singular, while _vaatas_ does (has both `ps3` and `sg`).


#### Using `NOT_MEMBERSHIP` mode


The fourth condition mode, `NOT_MEMBERSHIP`, matches nodes whose `upostag` value is neither `V` nor `S`. This will match all nodes except verbs and nouns.


````python
# Pattern definition using NOT_MEMBERSHIP condition mode
v_and_s_pattern_not_membership = PathPattern(
    name="not_membership_verb_and_noun_pattern",
    node_steps=(
        NodeConstraint(
            role="not_verb_or_noun",
            attribute_conditions={
                "upostag": ValueCondition(
                    mode=ConditionMode.NOT_MEMBERSHIP, value=["V", "S"]
                )
            },
        ),
    ),
    edge_steps=(),
)
````

All nodes with `upostag` not equal to `V` or `S` are matched.

### Defining a custom decorator for the tagger


`DepChainTagger` is designed to be flexible and extensible. If you want to add custom processing logic to the matches produced by the tagger, you can define a custom decorator function and pass it to the tagger's `decorate_match` parameter. This function will be called for each match, allowing you to modify the match's payload before it is added to the output layer.

In the example below, we define two custom decorator functions: one that adds a new attribute `S_is_person` indicating whether the node filling the `S` role is a person (using WordNet hypernym matching), and another that filters such matches out by returning `None` if the `S` node is a person. We then create two tagger instances with the same pattern but different decorators to see how their outputs differ.


````python
from estnltk.wordnet import Wordnet

wn = Wordnet()
````

````python
def is_person(word: str) -> bool:
    """
    Check if the given word is a person by looking up its synsets in WordNet and checking if any of them have "elusolend.n.01" as a hypernym (directly or indirectly).

    ## Attributes
    - `word` (str): The word to check.

    ## Returns
    - `bool`: `True` if the word is a person, `False` otherwise.
    """
    synsets = wn[word]
    if not synsets:
        return False

    # Recursively check if any hypernym of the synsets is "elusolend.n.01"
    def has_person_hypernym(synset):
        if synset.name == "elusolend.n.01":
            return True
        for hypernym in synset.hypernyms:
            if has_person_hypernym(hypernym):
                return True
        return False

    return any(has_person_hypernym(synset) for synset in synsets)
````

````python
def decorator_add_flag(
    text: estnltk.Text, base_span: dict, annotation: dict
) -> dict | None:
    """
    Example of an annotation decorator function that checks if the second token in the matched pattern (the one with role "S") is a person according to WordNet, and if so, adds a flag to the annotation payload.

    ## Attributes
    - `text` (estnltk.Text): The text object being processed.
    - `base_span` (dict): A dictionary mapping role names to token spans that matched the pattern. For example, if the pattern has node constraints with roles "A", "B", and "S", the base_span might look like `{"A": token_span_a, "B": token_span_b, "S": token_span_s}`.
    - `annotation` (dict): The annotation dictionary that is being constructed for this match. This can be modified by the decorator to add additional information or to filter out the annotation by returning None.

    ## Returns
    - `dict | None`: The modified annotation dictionary to be used for the annotation, or `None` if the annotation should be filtered out (i.e., not created). In this example, we always return an annotation dictionary, but we add an "S_is_person" flag if the second token is a person according to WordNet.
    """
    annotation_payload = (
        annotation.copy()
    )  # Start with the original annotation payload, and we can modify it as needed.
    # Extract the text of the second token in the chain (the one with role "S")
    second_token_span = base_span.get("S")
    if second_token_span is not None:
        second_token_lemma = second_token_span.lemma
        if is_person(second_token_lemma):
            annotation_payload["S_is_person"] = (
                True  # Optionally add a flag to the annotation payload indicating that the second token is a person
            )
            return annotation_payload
    return annotation_payload
````

````python
def decorator_filter_out(
    text: estnltk.Text, base_span: dict, annotation: dict
) -> dict | None:
    """
    Example of an annotation decorator function that filters out the annotation if the second token in the matched pattern (the one with role "S") is a person according to WordNet by returning None.

    ## Attributes
    - `text` (estnltk.Text): The text object being processed.
    - `base_span` (dict): A dictionary mapping role names to token spans that matched the pattern. For example, if the pattern has node constraints with roles "A", "B", and "S", the base_span might look like `{"A": token_span_a, "B": token_span_b, "S": token_span_s}`.
    - `annotation` (dict): The annotation dictionary that is being constructed for this match. This can be modified by the decorator to add additional information or to filter out the annotation by returning None.

    ## Returns
    - `dict | None`: The modified annotation dictionary to be used for the annotation, or `None` if the annotation should be filtered out (i.e., not created). In this example, we return `None` if the second token is a person in WordNet, effectively filtering out any matches where it is.
    """
    annotation_payload = (
        annotation.copy()
    )  # Start with the original annotation payload, and we can modify it as needed.
    # Extract the text of the second token in the chain (the one with role "S")
    second_token_span = base_span.get("S")
    if second_token_span is not None:
        second_token_lemma = second_token_span.lemma
        if is_person(second_token_lemma):
            return None  # Filter out this annotation by returning None if the second token is a person
    return annotation_payload
````

Let's define a pattern that finds the "lendurist abikaasa" phrase.


````python
pattern = PathPattern(
    name="noun_st_noun",
    node_steps=(
        NodeConstraint(
            role="S_el",
            attribute_conditions={
                "upostag": ValueCondition(
                    mode=ConditionMode.MEMBERSHIP, value=["S", "A"]
                ),
                "deprel": ValueCondition(mode=ConditionMode.REGEX, value=".*mod.*"),
            },
            nested_attribute_conditions={
                "feats": NestedValueCondition(
                    mode=ConditionMode.EXACT, required={"sg": "sg", "el": "el"}
                )
            },
        ),
        NodeConstraint(
            role="S",
            attribute_conditions={
                "upostag": ValueCondition(mode=ConditionMode.EXACT, value="S"),
                "lemma": ValueCondition(
                    mode=ConditionMode.REGEX, value=".*(?<!mine)$"
                ),  # match any lemma that does not end with "mine" (to exclude verbal nouns)
            },
            nested_attribute_conditions={
                "feats": NestedValueCondition(
                    mode=ConditionMode.EXACT, required={"sg": "sg", "nom": "nom"}
                )
            },
        ),
    ),
    edge_steps=(
        EdgeConstraint(
            direction=DirectionMode.UP,
            min_hops=1,
            max_hops=3,
        ),
    ),
)
````

Now, let's instantiate the tagger with this pattern and the decorator that adds the `S_is_person` attribute, and see how it works on our sample text.


````python
dc_tagger = DepChainTagger(
    patterns=(pattern,),
    output_attributes=(
        "S_is_person",
    ),  # Specify which attributes from the annotation payload should be included in the output annotations
    include_pattern_constraints=False,
    annotation_decorator=decorator_add_flag,
)
````

````python
sample_text = "Lendurist abikaasa vaatas laeva."
text_obj = estnltk.Text(sample_text)
text_obj.tag_layer(
    "morph_extended"
)  # StanzaSyntaxTagger expects morph layer to be present
stanza_syntax_tagger.tag(text_obj)
# Run the tagger
dc_tagger.tag(text_obj);
````

The output layer contains one match for the pattern with the `S_is_person` attribute added by the custom decorator.

### _Deduplication modes_


Deduplication is primarily a defensive and future-facing feature. For the current sentence-tree matcher identical role→token assignments are rare, but duplicates can still arise in other situations (for example: duplicate lower-level inputs, intentionally repeated patterns in tests, non-tree graphs with multiple traversal paths, or when matches are merged across runs).

The available modes give you control over how aggressively to collapse similar matches:

- `none`: keep all matches (useful for exhaustive output or debugging).
- `exact`: collapse only fully identical matches (same role assignments and same `matched_text`).
- `role_based`: collapse matches that assign the same token IDs to roles (ignores `matched_text`).

# DepChildTagger


`DepChildTagger` is a specialised version of `DepChainTagger` that focuses on matching child nodes in dependency trees. It enforces that all edges in the patterns are directed downwards (from parent to child), which simplifies the matching process and ensures that only child relationships are considered.

**When to use each one?**

You should use `DepChildTagger` when your linguistic patterns strictly look at descendants (e.g., finding a verb and its direct objects/subjects). It will be faster and less resource-intensive. Use `DepChainTagger` when you need to look upwards (e.g., finding the governor of a node) or when the anchor is in the middle of a pattern.


````python
sample_text = "Väga ilmekas on lõik lastega, noore lese kirjast."
text_obj = estnltk.Text(sample_text)
text_obj.tag_layer("morph_extended")
stanza_syntax_tagger.tag(text_obj)
display(text_obj.stanza_syntax)
````

![Syntax tree](example_syntax_tree.png)

This illustration shows a syntax tree with a root node and its children. `DepChildTagger` is designed to match patterns that traverse downwards from a parent node to its child nodes. In contrast, `DepChainTagger` can traverse both upwards and downwards, allowing for more complex patterns that may involve ancestors or siblings in the tree.


Let's look at an example of mix-direction patterns that are valid for `DepChainTagger` but not for `DepChildTagger`. In this example, we have a pattern that starts at a leaf node, goes up to the root, and then down to the root's immediate child. This pattern is valid for `DepChainTagger` because it allows upward traversal, but it is invalid for `DepChildTagger` because it requires moving up the tree.


````python
pattern = PathPattern(
    name="noun_st_noun",
    node_steps=(
        NodeConstraint(
            role="S_el",
            attribute_conditions={
                "upostag": ValueCondition(
                    mode=ConditionMode.MEMBERSHIP, value=["S", "A"]
                ),
                # "deprel": ValueCondition(mode=ConditionMode.REGEX, value=".*mod.*"),
            },
            nested_attribute_conditions={
                "feats": NestedValueCondition(
                    mode=ConditionMode.EXACT,
                    required={"sg": "sg", "el": "el"},
                    allow_extra_keys=True,
                )
            },
        ),
        NodeConstraint(
            role="A",
            attribute_conditions={
                "upostag": ValueCondition(mode=ConditionMode.EXACT, value="A"),
            },
        ),
        NodeConstraint(
            role="S",
            attribute_conditions={
                "upostag": ValueCondition(mode=ConditionMode.EXACT, value="S"),
                "lemma": ValueCondition(mode=ConditionMode.REGEX, value=".*(?<!mine)$"),
            },
            nested_attribute_conditions={
                "feats": NestedValueCondition(
                    mode=ConditionMode.EXACT,
                    required={"sg": "sg", "nom": "nom"},
                    allow_extra_keys=True,
                )
            },
        ),
    ),
    edge_steps=(
        EdgeConstraint(
            direction=DirectionMode.UP,
            min_hops=2,
            max_hops=2,
        ),
        EdgeConstraint(
            direction=DirectionMode.DOWN,
            min_hops=1,
            max_hops=1,
        ),
    ),
)
````

Now, let's instantiate the tagger with this pattern and see how it works on our sample text.


````python
dc_tagger = DepChainTagger(patterns=(pattern,))
````

````python
sample_text = "Väga ilmekas on lõik lastega, noore lese kirjast."
text_obj = estnltk.Text(sample_text)
text_obj.tag_layer("morph_extended")
stanza_syntax_tagger.tag(text_obj)

# Run the tagger
dc_tagger.tag(text_obj);
````

Let's try to instantiate a `DepChildTagger` with the same pattern and see what happens.


````python
cd_tagger = DepChildTagger(patterns=(pattern,))
````

The tagger raises a `ValueError` because the pattern includes an upward edge, which is not allowed in `DepChildTagger`. This demonstrates the restriction that `DepChildTagger` imposes on patterns, requiring them to include only downward edges.

Now, let's create a valid pattern for `DepChildTagger` that only includes downward edges. We will define a pattern that matches a node and its immediate child.


````python
pattern = PathPattern(
    name="noun_st_noun",
    node_steps=(
        NodeConstraint(
            role="S_gen",
            attribute_conditions={
                "upostag": ValueCondition(
                    mode=ConditionMode.MEMBERSHIP, value=["S", "A"]
                ),
                # "deprel": ValueCondition(mode=ConditionMode.REGEX, value=".*mod.*"),
            },
            nested_attribute_conditions={
                "feats": NestedValueCondition(
                    mode=ConditionMode.EXACT,
                    required={"sg": "sg", "gen": "gen"},
                    allow_extra_keys=True,
                )
            },
        ),
        NodeConstraint(
            role="S",
            attribute_conditions={
                "upostag": ValueCondition(mode=ConditionMode.EXACT, value="S"),
                "lemma": ValueCondition(mode=ConditionMode.REGEX, value=".*(?<!mine)$"),
            },
            nested_attribute_conditions={
                "feats": NestedValueCondition(
                    mode=ConditionMode.EXACT,
                    required={"sg": "sg", "el": "el"},
                    allow_extra_keys=True,
                )
            },
        ),
    ),
    edge_steps=(
        EdgeConstraint(
            direction=DirectionMode.DOWN,
            min_hops=1,
            max_hops=1,
        ),
    ),
)
````

Now, let's instantiate the tagger with this pattern and see how it works on our sample text.


````python
cd_tagger = DepChildTagger(patterns=(pattern,))
````

From the example above, the difference between `DepChainTagger` and `DepChildTagger` might not be immediately apparent because the pattern we defined for `DepChildTagger` can be used in `DepChainTagger` as well. However, the key point is that `DepChildTagger` does not have any reverse traversals or multi-directional checks, and thus has a smaller search space. This makes it more efficient for patterns that only need to look at child nodes and their relationships, thus being usually faster and less resource-intensive than `DepChainTagger` for such use cases. In contrast, `DepChainTagger` is more flexible and can handle a wider range of patterns, including those that require upward traversal or more complex relationships between nodes.
